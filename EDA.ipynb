# `Visulaizations`

### `Total orders count by order status.`

```python
# Order_item_table

order_table.head(3)
```

```python
orders_count = order_table['order_status'].value_counts()

orders_count
```

```python
# Bar chart

ax = orders_count.plot(kind = 'bar',figsize=(14,7))

ax.bar_label(ax.containers[0])

plt.title('Total orders count')

plt.xlabel('Order status')
plt.ylabel('Orders count')

plt.style.use('seaborn-v0_8')

plt.xticks(rotation = 0)

plt.tight_layout()

plt.show()
```

#### `Insight `

##### `Most orders are delivererd and after that most are shipped already and unfortunately 602 orders are cancelled.`

## `Montly order trend`

```python
order_table.columns
```

```python
total_order_by_months = order_table.groupby(order_table['order_purchase_timestamp'].dt.to_period('M'))['order_id'].count()

total_order_by_months
```

```python
total_order_by_months.plot(kind= 'line', figsize = (13,6), marker = '*')

plt.title('Orders by Months')

for x,y in total_order_by_months.items():
    plt.text(x,y, f'{y}', ha = 'center',va = 'bottom', fontsize = 10)

plt.xlabel('Months')
plt.ylabel('Orders')

plt.style.use('seaborn-v0_8')

# plt.tight_layout()

plt.show()
```

## `Insight :`

#### `Maximum order were recived during oct 2017 to novemeber 2017.`
#### `we can see a sharp drop of orders after july 2018 or agaust 2018.`

# `Top 10 Cities by Customers`

```python
customer_table.head(3)
```

```python
top_10_city_customers = customer_table.groupby('customer_city')['customer_id'].count().sort_values(ascending=False).head(10)

top_10_city_customers
```

```python
ax = top_10_city_customers.plot(kind = 'barh', figsize=(13,7))

ax.bar_label(ax.containers[0])

ax.invert_yaxis()

plt.title('Top City by customers')

plt.xlabel('Customers count')
plt.ylabel('City')

plt.style.use('ggplot')

plt.tight_layout()

plt.show()
```

## `Top 10 Product categories`

```python
# Overview

product_table.head(3)
```

```python
top_10_product_categories = product_table.groupby('product_category_name')['product_id'].count().sort_values(ascending=False).head(10)

top_10_product_categories
```

```python
ax = top_10_product_categories.plot(kind = 'barh',figsize= (13,6))

ax.bar_label(ax.containers[0])

ax.invert_yaxis()

plt.title('Top 10 Product Categories')

plt.xlabel('Products Count')
plt.ylabel('Product Categories')

plt.style.use('seaborn-v0_8-whitegrid')

plt.tight_layout()

plt.show()
```
